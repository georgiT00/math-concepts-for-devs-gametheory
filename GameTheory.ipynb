# Game Theory

# 1. Introduction & Definition

Game theory is a mathematical representation, which observes a person's decision, knowing that other people's decisions also affect the outcome of a certain situation.

"A Beautiful Mind(2001)", is movie in which I was introduced to the game theory. The movie stars Russell Crowe playing as John Nash, an American mathematician, who has made a significant contribution to the game theory. Later in this article, we will explore Nash's Equilibrium.
